## multioutput

有三种可选择的取值：

uniform_average (default) :所有输出的误差均以统一的权重取平均值。

raw_values:在多输出输入的情况下返回完整的误差值。

数组：要求和输入形状相同。

### 一维两个元素

```python
import numpy as np
from sklearn.metrics import mean_squared_error as MSE

a = np.random.rand(2)
b = np.random.rand(2)
print(a)
print(b)
```

```python
res = MSE(a,b)
res
```

```python
res = MSE(a,b,multioutput='uniform_average')
res
```

```python
res = MSE(a,b,multioutput='raw_values')
res
```

```python
res = MSE(a,b,multioutput=[1.0])
res
```

### 二维两个元素

```python
import numpy as np
from sklearn.metrics import mean_squared_error as MSE

np.random.seed(1998)
a = np.random.rand(1,2)
b = np.random.rand(1,2)
print(a)
print(b)
```

```python
res = MSE(a,b)
res
```

```python
res = MSE(a,b,multioutput='uniform_average')
res
```

```python
res = MSE(a,b,multioutput='raw_values')
res
```

```python
np.mean(res)
```

```python
res = MSE(a,b,multioutput=[0.5, 0.5])
res
```

```python
res = MSE(a,b,multioutput=[1.0, 1.0])
res
```
