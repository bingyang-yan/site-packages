## 笔记

datetime包含4类时间数据，分别是date，datetime，time，timedelta，都是非字符串

前3类时间数据，如果要输出为字符串，主要有两种方式：d.__str__() 或者 d.__format__('%Y-%m-%d %H:%M:%S')

timedelta为时间差，有3个属性：days，seconds，microseconds，注意3个系统是独立的，days独立做差，seconds独立做差

### datetime类

#### 获取当前日期时间

```python
import datetime

b = datetime.datetime.now()

print(b)
print(b.year)
print(b.month)
print(b.day)
print(b.hour)
print(b.minute)
print(b.second)
print(b.microsecond)
```

#### 获取指定日期时间

```python
h = datetime.datetime(2022,2,7,12,58,30)
print(type(h))
print(h)
```

#### 比较日期时间大小

```python
h = datetime.datetime(2022,2,7,12,58,30)
i = datetime.datetime(2023,2,7,12,58,30)

print(h.__lt__(i))
```

#### 日期时间的字符串输出

```python
h = datetime.datetime(2022,2,7,12,58,30)

print(h.__str__())

print(h.__format__('%Y-%m-%d %H:%M:%S'))
```

### time类

#### 获取指定时间

#### 时间的字符串输出

```python
d = datetime.time(12,30,21)

print(d.__str__())
print(d.__format__('%H:%M:%S'))
```

### timedelta类

这是一个日期时间差，只有days，seconds，microseconds三个属性

**注意三个类是独立的系统**

#### 计算日期差

```python
e = datetime.date(2021,2,7)
f = datetime.date(2022,2,7)
g = e.__rsub__(f)

print(type(g))
print(g)
print(g.days)
print(g.seconds)
```

```python
g.total_seconds()
```

#### 计算日期时间差

```python
h = datetime.datetime(2022,2,7,12,58,30)
i = datetime.datetime(2022,2,8,12,59,30)
j = i.__sub__(h)

print(type(j))
print(j)
print(j.days)
print(j.seconds)
```

#### 将日期差转换成总秒数

```python
h = datetime.datetime(2022,2,7,12,58,30)
i = datetime.datetime(2022,2,8,12,59,30)

j = i - h
j.total_seconds()
```

#### 创建一个timedelta

```python
a = datetime.timedelta(days=2)
now = datetime.datetime.now()
now+a
```
