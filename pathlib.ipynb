## 常用

### cwd()

```python
from pathlib import Path

Path.cwd()
```

### parent

```python
from pathlib import Path

Path.cwd().parent
```

### iterdir()

```python
[i for i in Path.cwd().iterdir()]
```

### glob

```python
from pathlib import Path

[i for i in Path.cwd().glob('*a*')]
```
