## 介绍

这是一个插值函数

## 举例

```python
import numpy as np

a = np.arange(0, 10)
b = np.exp(-a / 3.0)

print(a)
print(b)
```

```python
from scipy.interpolate import interp1d

f = interp1d(a, b, bounds_error=False, fill_value='extrapolate')

c = f([11])
c

import matplotlib.pyplot as plt
plt.plot(f(np.arange(1, 15, 1)))
```
