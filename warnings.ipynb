## 显示警告

```python
import warnings

warnings.simplefilter('always')
if True:
    warnings.warn('yby1')
if False:
    warnings.warn('yby2')
    
warnings.simplefilter('ignore')
if True:
    warnings.warn('yby3')
if False:
    warnings.warn('yby4')
    
warnings.simplefilter('always')
if True:
    warnings.warn('yby5')
if False:
    warnings.warn('yby6')
    
with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    warnings.warn('yby7')
    
warnings.warn('yby8')
    
with warnings.catch_warnings():
    warnings.simplefilter('always')
    warnings.warn('yby9')
    
warnings.warn('yby10')
```

## 笔记

warnings.simplefilter用来设置大环境的action，设置完成后在此环境一直持续

with warnings.catch_warnings()用来设置小环境的action，不管大环境的action如何，小环境中的action只在小环境生效，并且不会改变大环境的action
