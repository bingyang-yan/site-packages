## squeeze & expend_dims

### squeeze

**如果不指定axis，是压缩所有的1**

```python
import numpy as np

a = np.empty(shape=(1,2,1,3))
print('a.shape:', a.shape)
print(a)

a = np.squeeze(a)
print('a.shape:', a.shape)
print(a)

a = np.empty(shape=(1,2,1,3))
print('a.shape:', a.shape)
print(a)

a = np.squeeze(a, 0)
print('a.shape:', a.shape)
print(a)
```

### expand_dims

```python
import numpy as np

a = np.empty(shape=(2, 3))
print('a.shape:', a.shape)
print(a)

a = np.expand_dims(a, 0)
print('a.shape:', a.shape)
print(a)
```

## 数组拼接

### stack

**只能拼接两个相同维度的数组**

```python
import numpy as np
array = np.empty([2,2])
print('empty', array)
new = np.random.rand(2,2)
print('new', new)
array = np.stack((array, new))
print('array', array)
```

### insert

**注意，一定要指定axis=0，默认是flattern**

**insert idx**

insert是在idx之前插入，所以idx设置成1是第一个不变，然后一直在第二位实现插入，从第二位开始是倒叙

如果设置成-1，是全部倒叙

如果设置成a.shape[0]，能够实现类似list.append的功能

```python
import numpy as np

a = np.array([])
for i in range(5):
    a = np.insert(a, 0, i)
a
```

```python
import numpy as np

a = np.array([])
for i in range(5):
    a = np.insert(a, -1, i)
a
```

```python
import numpy as np
a = np.empty(shape=(1, 4, 3, 2))
b = np.random.rand(4,3,2)
c = np.random.rand(4,3,2)

print('a.shape:',a.shape)
print(a)
print('b.shape:',b.shape)
print(b)
print('c.shape:',c.shape)
print(c)


print('a + b:')
a = np.insert(a, a.shape[0], b, axis=0)
print(a)
print('a + b + c:')
a = np.insert(a, a.shape[0], c, axis=0)
print(a)
```

### append

发现这种方法会减少一个维度

```python
import numpy as np

a = np.array([])

for i in range(5):
    i = np.array([i, i])
    print(i)
    a = np.append(a, i, axis=0)
    
a
```

append也会减少一个维度，要求要append的数组包含一个最外面的壳，append的是其中的元素

```python
import numpy as np

a = np.empty(shape=(5,2))
print(a)
for i in range(5):
    i = np.array([[i,i]])
    a = np.append(a, i, axis=0)
a
```

```python
import numpy as np
a = np.empty(shape=(4, 3, 2))
b = np.random.rand(4,3,2)
c = np.random.rand(4,3,2)

a = np.extend_dims(a)
print('a.shape:',a.shape)
print(a)
print('b.shape:',b.shape)
print(b)
print('c.shape:',c.shape)
print(c)


print('a + b:')
a = np.append(a, b, axis=0)
print(a)
print('a + b + c:')
a = np.append(a, c, axis=0)
print(a)
print(a.shape)
```

### expand_dim + append

**不在循环外给定一个空数组是无法完成的**

```python
import numpy as np

a = np.empty((2,3))
b = np.empty((2,3))
data_list = [a, b]
print(data_list)

data = np.empty(shape=(0,2,3))
for i in data_list:
    i = np.expand_dims(i, 0)
    data = np.append(data ,i, 0)
    
print('data.shape:',data.shape)
print(data)
```

## append vs empty time-consuming

```python
import numpy as np
import time

total=1e6
total = int(total)

time1=time.time()
data=np.array([])
for i in range(total):
    data=np.append(data,i)
m1=max(data)
time2=time.time()

time3=time.time()
data=np.empty(total)
for i,v in enumerate(range(total)):
    data[i]=v
m1=max(data)
time4=time.time()

print(time2 - time1)
print(time4 - time3)
```

## 总结

### 数据存储方式

numpy array是绝对不可以存储元素形状不同的高维数据。如果要存储这种类型的数据，只能先flatten，再反flatten。

append数据过程中，最好使用列表的append，而且列表的append不要求元素的类型相同，如果使用np.append不指定axis会默认flatten。

### 数组运算

点乘要求shape相同，矩阵乘法要求ij*jk

数组的点乘：* 

数组的矩阵乘法：dot（生成数组） multiply（生成矩阵）

矩阵的点乘：multiply

矩阵的矩阵乘法：* dot

总结：dot都是矩阵乘法，数组的点乘用*， 矩阵的点乘用multiply

### 创建空数组

使用[]创建不需要提前指定数组元素的维度，使用np.empty创建需要提前指定数组元素的维度。

使用[] append可以添加不同维度的数组，但是需要在进行np.array转换的时候指定dtype=object。**缺点是object不能进行flatten**
