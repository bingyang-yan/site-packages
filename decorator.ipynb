## 函数装饰器

```python
def dec(dec_para):
    def outwarpper(func):
        def warpper(*args, **kargs):
            print('dec_para', dec_para)
            return func(*args, **kargs)
        return warpper
    return outwarpper
```

```python
@dec('yby')
def fun(a,b):
    print('a', a)
    print('b', b)
```

```python
fun(1,2)
```

## 类装饰器

```python
class dec():
    
    def __init__(self, dec_para):
        self.dec_para = dec_para
        
    def __call__(self, func):
        def warpper(*args, **kargs):
            print(self.dec_para)
            return func(*args, **kargs)
        
        return warpper
```

```python
@dec('yby')
def fun(a,b):
    print('a', a)
    print('b', b)
```

```python
fun(2,3)
```
