## 笔记

### namedtuple

**如果作为函数参数，禁用此变量类型，因为每次读取log的时候都需要指定自定义的那个namedtuple，尝试单独传入key**

谨慎使用该类型，因为该类型不可被pickle序列化保存，需要将指定的namedtuple变量指定到我编写的类中，pickle序列化才不会出错。

就算保存成功，如果读取时没有指定当时保存时使用的namedtuple变量，也会读取失败。
