## 笔记

区别于datetime的4类时间数据，time只有2类时间数据：时间戳 和 元组， **元组是通用格式，也是转换的桥梁**

**时间戳---元组---格式化日期时间，元组作为中间桥梁，如果要做时间戳和格式化日期时间的转换，先转换为元组**

## 功能

### 获取当前的时间戳

```python
import time

a = time.time()
print(a)
```

### 获取当前的元组

```python
import time

b = time.localtime()
print(b)
```

### 将时间戳转换为元组

```python
c = time.localtime(a)
print(c)
```

### 将元组转换为时间戳

```python
d = time.mktime(c)
print(d)
```

### 将元组转换为格式化日期时间（字符串）

```python
e = time.strftime('%Y-%m-%d %H:%M:%S', c)
print(type(e))
print(e)
```

### 将格式化日期时间（字符串）转换为元组

```python
f = time.strptime(e, '%Y-%m-%d %H:%M:%S')
print(type(f))
print(f)
```
