## 笔记

**在操作文件夹时：**

进行复制或者移动时，比如f1中的文件夹1，复制或移动到f2时，要求f2中没有文件夹1，否则就会报错。

**在操作文件时：**

在进行复制时，比如f1文件夹1中的1.txt，复制到f2文件夹1中的1.txt，有两种写法，目标文件路径可以写f2中的文件夹1，也可以写f2文件夹1中的1.txt

在进行移动时，比如f1文件夹1中的1.txt，复制到f2文件夹1中的1.txt，目标文件路径只可以定位到文件，如果定位到f2中的文件夹1就会报错

### 配置环境

```python
import os
import shutil

root = r"C:\Users\CK\Desktop\new"

def config(root):    

    if not os.path.exists(root):
        os.mkdir(root)

    folder_level_1_num = 2
    folder_level_1 = []

    for i in range(folder_level_1_num):
        folder_level_1.append(os.path.join(root,'f'+str(i+1)))
    for i in folder_level_1:    
        if not os.path.exists(i):
            os.mkdir(i)

    folder_level_2_num = 4
    folder_level_2 = []

    for i in range(folder_level_2_num):
        for j in folder_level_1:
            folder_level_2.append(os.path.join(j,str(i+1)))
    for i in folder_level_2:
        if not os.path.exists(i):
            os.mkdir(i)
        with open(os.path.join(i,str(os.path.split(i)[1])+'.txt'), 'w+') as fp:
            fp.write('111')

    shutil.rmtree(folder_level_2[5])
    shutil.rmtree(folder_level_2[6])
    
    os.mkdir(r'C:\Users\CK\Desktop\new\f1\2\4')
    
    with open(r'C:\Users\CK\Desktop\new\f1\1\5.txt', 'w+') as fp:
        fp.read()

    for i in (folder_level_2):
        print(i)

    for r,d,f in os.walk(root):
        for i in d:
            print('dir:', os.path.join(r, i))
        for j in f:
            print('file:', os.path.join(r, j))   
```

```python
config(root)
```

```python
shutil.rmtree(root)
```
