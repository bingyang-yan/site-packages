## 基础流程

```python
import sys
from PyQt5.Qt import *

app = QApplication(sys.argv)

window = QWidget()
window.setWindowTitle('yby test')
window.resize(800, 300)
window.move(1000, 200)

window.show()
sys.exit(app.exec_())
```

## 按钮控件

### QPushButton

http://m.biancheng.net/view/9416.html

```python
app = QApplication(sys.argv)
window = Window()

a = QPushButton(window)
a.setText('test')
a.resize(200, 200)
a.move(100,200)

window.show()
sys.exit(app.exec_())
```

### QCommandLinkButton

```python
app = QApplication(sys.argv)
window = Window()

a = QCommandLinkButton(window)
a.setText('test')

window.show()
sys.exit(app.exec_())
```

### QRadioButton

```python
app = QApplication(sys.argv)
window = Window()

a = QRadioButton(window)
a.setText('test')

window.show()
sys.exit(app.exec_())
```

### QCheckBox

```python
app = QApplication(sys.argv)
window = Window()

a = QCheckBox(window)
a.setText(['test1', 'test2'])

window.show()
sys.exit(app.exec_())
```
