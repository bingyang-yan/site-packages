### 定义目标函数

```python
import numpy as np

def schaffer(p):
    '''
    This function has plenty of local minimum, with strong shocks
    global minimum at (0,0) with value 0
    https://en.wikipedia.org/wiki/Test_functions_for_optimization
    '''
    x1, x2 = p
    part1 = np.square(x1) - np.square(x2)
    part2 = np.square(x1) + np.square(x2)
    return 0.5 + (np.square(np.sin(part1)) - 0.5) / np.square(1 + 0.001 * part2)
```

### 可视化目标函数

```python
import matplotlib.pyplot as plt

x = np.arange(0,100,0.1)
x = x.reshape(-1,2)
y = []
for i in x:
    y.append(schaffer(i))
plt.plot(y)
```

### 遗传算法及属性

```python
from sko.GA import GA

ga = GA(func=schaffer, n_dim=2, size_pop=50, max_iter=800, prob_mut=0.001, lb=-1, ub=1, precision=1e-7)
best_x, best_y = ga.run()
print('best_x:', best_x, '\n', 'best_y:', best_y)
```

## 函数参数

### func

编写目标函数，目标函数的最小值对应的输入就是GA算法的最优解。（不清楚负值最小是最优解还是0是最优解，请使用abs(output-label)）

### n_dim

目标函数中输入参数的维度

### size_pop

种群大小，不允许为奇数，default is 50

### max_iter

最大迭代次数, default is 200

### prob_mut

变异率, default is 0.001

### lb

输入参数值的下限

### ub

输入参数值的上限

### constraint_eq

使用线性方程对输入值的某种关系进行限制，形式是列表，列表的元素是函数，函数的值为0

### constraint_ueq

使用非线性方程对输入值的某种关系进行限制，形式是列表，列表的元素是函数，函数的值为<=0

### precision

精度,default is 1e-7

### early_stop 

default is None
