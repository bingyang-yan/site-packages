## 基础知识

python中的list，dict， set都是可变对象，值很容易被新指向他的变量改变。

## 举例

```python
y = [1,2]
z = y
z[0] = 3
y
```

```python
y = {}
y['a'] = 'a'
z = y
z['a'] ='b'
y['a']
```

## copy

```python
import copy

y = [1,2]
z = copy.deepcopy(y)
z[0] = 3
y
```

```python
import copy

y = {}
y['a'] = 'a'
z = copy.deepcopy(y)
z['a'] ='b'
y['a']
```

## 函数

```python
def test1(the_list):
    
    the_list[1] = 'yby'
    
    return the_list
```

```python
a = [1,2,3]
b = test1(a)
print(a,b)
```

```python
def test2(the_list):
    
    from copy import deepcopy
    
    the_list = deepcopy(the_list)
    the_list[1] = 'yby'
    
    return the_list
```

```python
a = [1,2,3]
b = test2(a)
print(a,b)
```
