```python
"""
__author__ = 'linlei'
__project__:mayavi_demo
__time__:2021/12/11 9:30
__email__:"919711601@qq.com"
"""
import numpy as np
import h5py
from mayavi.mlab import pipeline
from mayavi import mlab



# read data
def read_h5(path):
    f = h5py.File(path,"r")
    data = f["/data"]
    return np.array(data)


# vision
def dataVisaliztion(data,index_x = 64,index_y = 64,index_z = 64,colorflag = "Greys"):
    mlab.figure(bgcolor=(1,1,1))    # background
    colorflag = colorflag   # colorbar
    pipeline.image_plane_widget(pipeline.scalar_field(data),
                                plane_orientation = "x_axes",
                                slice_index = index_x,
                                colormap = colorflag)
    pipeline.image_plane_widget(pipeline.scalar_field(data),
                                plane_orientation = "y_axes",
                                slice_index = index_y,
                                colormap = colorflag)
    mlab.pipeline.image_plane_widget(mlab.pipeline.scalar_field(data),
                                     plane_orientation = 'z_axes',
                                     slice_index = index_z,
                                     colormap = colorflag)
    # mlab.colorbar()
    # mlab.outline()
    mlab.show()

if __name__ == "__main__":
    path = "./fault_test.h5"
    data = read_h5(path = path)
    dataVisaliztion(data)
```

## y_mayavi

```python
def y_points3dd(x, y, z, value, opacity=0.9, cmap='jet', title='yyy'):
    
    '''
    plot scatter on mayavi
    
    PARAMETERS:
    -----------
    x        - x array
    y        - y array
    z        - z array
    value    - value at coordinate(x, y, z)
    opacity  - default is 1, optional: [1, 0.9]
    cmap     - color map 
    title    - default is None
    '''

    from mayavi import mlab
    
    mlab.points3d(x, y, z, value, mode='cube', scale_mode='none', opacity=opacity, transparent=False, colormap=cmap)
    mlab.colorbar(orientation='vertical', nb_labels=6)

    mlab.title(title, height=0.9, size=1.5, color=(1,0,0))
    
    mlab.show()
```

## 方位角和仰角的设置

```python
from mayavi import mlab
import numpy as np

# 创建一些3D数据
x = [0, 1, 2, 3, 4]
y = [0, 1, 2, 3, 4]
z = [0, 1, 2, 3, 4]

# 默认视角
mlab.figure('Default View')
mlab.points3d(x, y, z, scale_factor=0.5)
mlab.show()

# 视角A: 方位角为0度，仰角为0度
mlab.figure('View A: Azimuth 0, Elevation 0')
mlab.points3d(x, y, z, scale_factor=0.5)
mlab.view(azimuth=0, elevation=0)
mlab.show()

# 视角B: 方位角为90度，仰角为0度
mlab.figure('View B: Azimuth 90, Elevation 0')
mlab.points3d(x, y, z, scale_factor=0.5)
mlab.view(azimuth=90, elevation=0)
mlab.show()

# 视角C: 方位角为0度，仰角为90度
mlab.figure('View C: Azimuth 0, Elevation 90')
mlab.points3d(x, y, z, scale_factor=0.5)
mlab.view(azimuth=0, elevation=90)
mlab.show()
```
