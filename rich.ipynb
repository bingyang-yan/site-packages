## rich vs tqdm

rich的优点是好看，打印输出进度条不会中断

rich的缺点是配置比较麻烦，而且会出现进度条闪烁的情况

tqdm的优点是经典，配置简单，bug很少，默认配置显示的信息很全

tqdm的缺点是打印输出进度条会中断，和rich相比不好看

总结，多线程推荐使用rich

## progress教程一

```python
import time
from rich.progress import Progress, TextColumn, BarColumn, TimeElapsedColumn, TimeRemainingColumn, SpinnerColumn, FileSizeColumn, DownloadColumn, ProgressColumn, TotalFileSizeColumn, TransferSpeedColumn
with Progress(TextColumn("[progress.description]{task.description}"),
                    SpinnerColumn(),
                    BarColumn(),
                    FileSizeColumn(),
                    TotalFileSizeColumn(),
                    DownloadColumn(),
                    TransferSpeedColumn(),
                    TextColumn("[progress.percentage]{task.percentage:>3.0f}%"),
                    TimeRemainingColumn(),
                    TimeElapsedColumn()) as progress:

    epoch_tqdm = progress.add_task(description="epoch progress", total=10)
    batch_tqdm = progress.add_task(description="batch progress", total=100)
    for ep in range(10):
        for batch in range(100):
            print("ep: {} batch: {}".format(ep, batch))
            progress.advance(batch_tqdm, advance=1)
            time.sleep(0.1)
        progress.advance(epoch_tqdm, advance=1)
        progress.reset(batch_tqdm)
```

## progress教程2

```python
from rich.progress import Progress,TextColumn,SpinnerColumn,BarColumn,FileSizeColumn,TotalFileSizeColumn
from rich.progress import DownloadColumn,TransferSpeedColumn,TextColumn,TimeRemainingColumn,TimeElapsedColumn
import time

p = Progress(
    TextColumn("[progress.description]{task.description}"),
    SpinnerColumn(),
    BarColumn(),
    #FileSizeColumn(),
    #TotalFileSizeColumn(),
    #DownloadColumn(),
    #TransferSpeedColumn(),
    TextColumn("[progress.percentage]{task.percentage:>3.0f}%"),
    TimeRemainingColumn(),
    TimeElapsedColumn(),
    transient=True,
    disable=False
)
taska = p.add_task(description='a', total=3)
taskb = p.add_task(description='b', total=3)
p.start()
for a in range(4):
    p.advance(taska)
    for b in range(4):
        p.advance(taskb)
        time.sleep(1)
    p.reset(taskb)
p.stop()
```

## y_rich

```python
def y_rich(iterator, desc='Processing', transient=False, verbose=True):
    
    from rich.progress import Progress,TextColumn,SpinnerColumn,BarColumn,FileSizeColumn,TotalFileSizeColumn
    from rich.progress import DownloadColumn,TransferSpeedColumn,TextColumn,TimeRemainingColumn,TimeElapsedColumn
    
    p = Progress(
        TextColumn("[progress.description]{task.description}"),
        SpinnerColumn(),
        BarColumn(),
        #FileSizeColumn(),
        #TotalFileSizeColumn(),
        #DownloadColumn(),
        #TransferSpeedColumn(),
        TextColumn("[progress.percentage]{task.percentage:>3.0f}%"),
        TimeRemainingColumn(),
        TimeElapsedColumn(),
        transient=transient,
        disable=not verbose
        )
    
    t = p.add_task(description=desc, total=len(iterator)-1)
    p.start()
    
    return p,t
```

```python
import time
p,t = y_rich([1,2,3,4], desc='yyyyy', verbose=False)
for i in range(4):
    time.sleep(1)
    p.advance(t)
p.stop()
```

```python
import time
from rich.progress import Progress

total_timesteps = 10

with Progress() as progress:
    task = progress.add_task('yby', total=total_timesteps)

    for _ in range(total_timesteps):
        time.sleep(1)
        progress.advance(task, 1)
```
